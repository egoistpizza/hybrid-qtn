# HybridQTN: one Kaggle training session

Runs `scripts/kaggle_session.py`: set-up, continuing the previous session, training until the
time budget, and collecting the results. Everything is saved under `/kaggle/working/hybrid-qtn`.

**Before the first session**
1. *Settings*: Accelerator **GPU T4 x2**, Internet **on**.
2. *Add-ons → Secrets*: add `WANDB_API_KEY` and attach it to this notebook (without it, W&B logs offline).
3. Set `REF`, `DATASET`, `TAG` and `QUEUE_OPTIONS` below.

**Every session**: *Save Version → Save & Run All (Commit)*. The browser can be closed; the run is followed in W&B.

**To continue after a session ends**: *Add Input → Your Work → Notebooks → this notebook*, pick the
**latest** version's output (remove any older one), then *Save & Run All* again. Unfinished runs resume from
their last finished epoch; finished runs are skipped. The same settings must be kept between sessions.

```python
REF = "feat/duts-predefined"   # branch or tag to train, e.g. "paper-v1"; the same for every session
DATASET = "duts"
TAG = "paper_v1"               # code version of the runs; change it whenever the code changes
QUEUE_OPTIONS = "--tiers 1"    # passed to scripts.run_queue, e.g. "--tiers 1 --seeds 42 43"
```
